```python
def polyline(n, length, angle):
    """Draws line segments with the given length and angle between them.
    
    n: integer number of line segments
    length: length of the line segments
    angle: angle between segments (in degrees)
    """    
    for i in range(n):
        forward(length)
        left(angle)
```

```python
from jupyturtle import penup, pendown

def jump(length):
    """Move forward length units without leaving a trail.
    
    Postcondition: Leaves the pen down.
    """
    penup()
    forward(length)
    pendown()
```

```python
def rhombus(side_length, angle):
    
    for _ in range(2):
        forward(side_length)
        left(angle)
        forward(side_length)
        left(180 - angle)
```

```python
def parallelogram(width, height, angle):
   
    for _ in range(2):
        forward(width)
        left(angle)
        forward(height)
        left(180 - angle)
```

```python

def rectangle(t, width, height):
    parallelogram(width, height, 90)
```

```python
import math

def triangle( radius, angle):

    y = radius * math.sin(angle * math.pi / 180.0)
    right_angle = 90 - angle
    base = 2 * y
    
    forward(radius)
    left(90 + right_angle)
    forward(base)
    left(90 + right_angle)
    forward(radius)
    left(180 - angle) # Devolve a tartaruga para a posição original

def draw_pie(n, radius):
   
    angle = 360.0 / n
    for _ in range(n):
        triangle(radius, angle / 2)
        left(angle)
```

```python
def arc(r, angle):
    
    arc_length = 2 * math.pi * r * angle / 360
    n = int(arc_length / 3) + 1
    step_length = arc_length / n
    step_angle = float(angle) / n
    
    for _ in range(n):
        forward(step_length)
        left(step_angle)

def petal(radius, angle):
  
    for _ in range(2):
        arc( radius, angle)
        left(180 - angle)

def flower(n, radius, angle):
    
    for _ in range(n):
        petal( radius, angle)
        left(360.0 / n)
```

```python
make_turtle()
flower( 7, 60, 60)
```
